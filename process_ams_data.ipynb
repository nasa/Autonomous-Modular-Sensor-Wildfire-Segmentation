```python
import AMS_dataset_processing
import dataset_utils
import dataset_patching
import glob
import tarfile
import os
import cv2
import tifffile
import numpy as np
```

This unzips the NASA AMS Data from https://asapdata.arc.nasa.gov/share/JBaculi/

```python
orig_files = glob.glob("./data/ams_data/**/*.tar")
for file in orig_files:
    print(file)
    tf = tarfile.open(file, "r:")
    tf.extractall("./data/ams_data/extracted_hdf_files")
    tf.close()
```

This processes the hdf files generated from the AMS sensor, reorients the images, separates them into single band grayscale images, crops them just to the range of the original hdf, and then enhances them using CLAHE

```python
hdf_files = []

for root, dirs, files in os.walk("./data/ams_data/extracted_hdf_files"):
    for file in files:
        if file.endswith(".hdf"):
             hdf_files.append(os.path.join(root, file))
                
print("Number of files: " + str(len(hdf_files)))

out_folder = './data/ams_data/processed_images_new/labelled_images/complete_images'
AMS_dataset_processing.extract_all_bands(hdf_files, out_folder, overwrite=True)
```

```python
def generate_sub_channel_images(bands, orig_folder, dest_folder, subscript):
    complete_images = glob.glob(f'{orig_folder}/*.tif')
    for img in complete_images:
        new_img = dataset_utils.load_tif_image(img, bands)
        print(new_img.shape)
        print(new_img.min())
        print(new_img.max())
        # new_img *= 500
        new_name = f'{dest_folder}/{subscript}_{os.path.basename(img)}'
        tifffile.imwrite(new_name, new_img.astype(np.uint8))
        
```
